```python
categorical_columns = ['marital', 'education', 'housing', 'loan']

for col in categorical_columns:
    print("\n", "="*40)
    print(col.upper())
    print("="*40)
    print(df[col].value_counts())
```

```python
categorical_columns = ['job', 'marital', 'education', 'housing', 'loan', 'contact', 'month', 'poutcome', 'y']

for col in categorical_columns:
    print("\n", "="*40)
    print(col.upper())
    print("="*40)
    print(df[col].unique())
```

## Age Distribution Analysis

A histogram was created to understand the distribution of customer ages.

### Business Insight

The age distribution helps identify the dominant customer segments and supports targeted marketing strategies.

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(9,6))

sns.histplot(data=df, x='age', bins=30)

plt.title("Customer Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Customers")

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.countplot(data=df, x='job')

plt.title("Customer Distribution by Job")
plt.xlabel("Job Category")
plt.ylabel("Number of Customers")

plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(data=df, x='y')

plt.title("Customer Subscription Distribution")
plt.xlabel("Subscription Status")
plt.ylabel("Number of Customers")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(data=df, x='balance')

plt.title("Customer Balance Outlier Analysis")
plt.xlabel("Balance")

plt.show()
```

## Business Insights

Based on the exploratory data analysis, the following business insights were identified:

- The dataset is clean, with no missing values or duplicate records, making it reliable for analysis.
- Most customers belong to the working-age population, representing the bank's primary customer segment.
- Customer distribution varies across job categories, indicating opportunities for segment-specific marketing.
- The majority of customers did not subscribe to the bank's term deposit, suggesting that current marketing campaigns have room for improvement.
- Housing and personal loan information can be used to identify cross-selling opportunities.
- Customer account balances show significant variation, including high-value customers who may be suitable for premium banking products.
- Customer demographics and financial behavior can help build targeted marketing strategies and improve campaign effectiveness.
